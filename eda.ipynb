```python
import pandas as pd

def validar_coordenadas(df):
    # 1. Verificar valores nulos
    nulos_lat = df['lat'].isnull().sum()
    nulos_lon = df['lon'].isnull().sum()
    
    # 2. Definir rangos válidos
    # Latitud: -90 a 90 | Longitud: -180 a 180
    fuera_rango = df[
        (df['lat'] < -90) | (df['lat'] > 90) | 
        (df['lon'] < -180) | (df['lon'] > 180)
    ]
    
    print("--- Reporte de Validación ---")
    print(f"Valores nulos en Latitud: {nulos_lat}")
    print(f"Valores nulos en Longitud: {nulos_lon}")
    print(f"Registros fuera de rango geográfico: {len(fuera_rango)}")
    
    if not fuera_rango.empty:
        print("\nEjemplos de datos erróneos:")
        print(fuera_rango[['lat', 'lon']].head())
    else:
        print("\n✅ Todos los valores están dentro de los rangos lógicos.")
    return fuera_rango

# Ejecución
fuera = validar_coordenadas(data)
```
